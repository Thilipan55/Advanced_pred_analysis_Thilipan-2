```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split

plt.style.use("ggplot")

RANDOM_STATE = 42
data = load_breast_cancer()

X = pd.DataFrame(
    data.data,
    columns=data.feature_names
)
y = (data.target == 0).astype(int)


print("Dataset Name : Breast Cancer Wisconsin (Diagnostic)")
print("Rows         :", X.shape[0])
print("Columns      :", X.shape[1])

print("\nTarget Names :", data.target_names)

print("\nPositive Class : Malignant (1)")
print("Negative Class : Benign (0)")

display(X.head())
X.info()


display(X.describe())
print("\n")

print("DATA AUDIT")
missing = X.isnull().sum()
display(missing)
duplicates = X.duplicated().sum()

print("Duplicate Rows :", duplicates)
constant = X.columns[X.nunique() <= 1]

print("Constant Columns")

print(list(constant))
audit = pd.DataFrame({"Datatype":X.dtypes, "Missing":X.isnull().sum(), "Unique":X.nunique(), "Minimum":X.min(), "Maximum":X.max()})

display(audit)
dataset_summary = pd.DataFrame({

    "Dataset":[
        "Breast Cancer Wisconsin"
    ],

    "Rows":[
        X.shape[0]
    ],

    "Unique Patients":[
        X.shape[0]
    ],

    "Predictors":[
        X.shape[1]
    ],
    "Positive Class":[
        "Malignant"
    ],
    "Positive %":[
        round((y==1).mean()*100,2)
    ],
    "Missing %":[
        round(
            X.isnull().sum().sum()/X.size*100,
            2
        )
    ],
    "Split Method":[
        "80-20 Stratified"
    ]

})

print("\nDataset Summary")

display(dataset_summary)
print("\n")
print("CLASS DISTRIBUTION")
class_counts = pd.Series(y).value_counts()
display(class_counts)
plt.figure(figsize=(6,4))
sns.countplot(x=pd.Series(y))

plt.xticks(
    [0,1],
    ["Benign","Malignant"]
)
plt.title("Class Distribution")
plt.xlabel("Class")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
selected = ["mean radius","mean texture","mean perimeter","mean area"]
eda = X[selected].copy()
eda["Target"] = y
for feature in selected:
    plt.figure(figsize=(6,4))
    sns.histplot(data=eda,x=feature,hue="Target",kde=True,stat="density",common_norm=False)
    plt.title(feature)
    plt.tight_layout()
    plt.show()

plt.figure(figsize=(12,10))

sns.heatmap(X.corr(),cmap="coolwarm")
plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,stratify=y,random_state=RANDOM_STATE)

print("\n")
print("TRAIN TEST SPLIT")
print("Training Samples :", X_train.shape[0])
print("Testing Samples  :", X_test.shape[0])
print("Training Features :", X_train.shape[1])
print("Testing Features  :", X_test.shape[1])
print("\nTraining Positive % :", round((y_train==1).mean()*100,2))
print("Testing Positive % :", round((y_test==1).mean()*100,2))
```

```python

from sklearn.dummy import DummyClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_validate

print("MODEL BUILDING")

scoring = {"accuracy":"accuracy","balanced_accuracy":"balanced_accuracy","precision":"precision","recall":"recall","f1":"f1","roc_auc":"roc_auc","average_precision":"average_precision"
}

print("\n")
print("DUMMY CLASSIFIER")

dummy = DummyClassifier(strategy="prior")
dummy_cv = cross_validate(dummy,X_train,y_train,cv=5,scoring=scoring,return_train_score=True)
dummy_summary = pd.DataFrame(dummy_cv)

print(dummy_summary)
print("\nAverage Scores")
display(dummy_summary.mean())

print("\n")
print("FULL DECISION TREE")

tree = DecisionTreeClassifier(criterion="gini",random_state=RANDOM_STATE)
tree_cv = cross_validate(tree,X_train,y_train,cv=5,scoring=scoring,return_train_score=True)
tree_summary = pd.DataFrame(tree_cv)
print(tree_summary)

print("\nAverage Scores")
display(tree_summary.mean())
print("\n")
print("OVERFITTING CHECK")

print("Training Accuracy :",round(tree_summary["train_accuracy"].mean(),3))
print("Validation Accuracy :",round(tree_summary["test_accuracy"].mean(),3))

gap = tree_summary["train_accuracy"].mean() - tree_summary["test_accuracy"].mean()

print("Train-Test Gap :", round(gap,3))
tree.fit(X_train,y_train)

print("\nTree Statistics")

print("Depth :", tree.get_depth())
print("Leaves :", tree.get_n_leaves())
print("Nodes :", tree.tree_.node_count)

print("\n")
print("PRE-PRUNED TREE")

pruned_tree = DecisionTreeClassifier(criterion="gini", max_depth=4, min_samples_leaf=5, random_state=RANDOM_STATE)
pruned_cv = cross_validate(pruned_tree, X_train, y_train,cv=5, scoring=scoring, return_train_score=True)

pruned_summary = pd.DataFrame(pruned_cv)
display(pruned_summary.mean())
pruned_tree.fit(X_train, y_train)

print("\nPruned Tree Statistics")
print("Depth :", pruned_tree.get_depth())
print("Leaves :", pruned_tree.get_n_leaves())
print("Nodes :", pruned_tree.tree_.node_count)
cv_table = pd.DataFrame({

"Model":[

"Dummy",

"Decision Tree",

"Pruned Tree"

],

"ROC-AUC Mean ± SD":[

f"{dummy_summary['test_roc_auc'].mean():.3f} ± {dummy_summary['test_roc_auc'].std():.3f}",

f"{tree_summary['test_roc_auc'].mean():.3f} ± {tree_summary['test_roc_auc'].std():.3f}",

f"{pruned_summary['test_roc_auc'].mean():.3f} ± {pruned_summary['test_roc_auc'].std():.3f}"

],

"PR-AUC Mean ± SD":[

f"{dummy_summary['test_average_precision'].mean():.3f} ± {dummy_summary['test_average_precision'].std():.3f}",

f"{tree_summary['test_average_precision'].mean():.3f} ± {tree_summary['test_average_precision'].std():.3f}",

f"{pruned_summary['test_average_precision'].mean():.3f} ± {pruned_summary['test_average_precision'].std():.3f}"

],

"Sensitivity Mean ± SD":[

f"{dummy_summary['test_recall'].mean():.3f} ± {dummy_summary['test_recall'].std():.3f}",

f"{tree_summary['test_recall'].mean():.3f} ± {tree_summary['test_recall'].std():.3f}",

f"{pruned_summary['test_recall'].mean():.3f} ± {pruned_summary['test_recall'].std():.3f}"

],

"Balanced Accuracy Mean ± SD":[

f"{dummy_summary['test_balanced_accuracy'].mean():.3f} ± {dummy_summary['test_balanced_accuracy'].std():.3f}",

f"{tree_summary['test_balanced_accuracy'].mean():.3f} ± {tree_summary['test_balanced_accuracy'].std():.3f}",

f"{pruned_summary['test_balanced_accuracy'].mean():.3f} ± {pruned_summary['test_balanced_accuracy'].std():.3f}"

],

"Complexity":[

"Baseline",

f"Depth={tree.get_depth()}",

f"Depth={pruned_tree.get_depth()}"

]

})

print("\n")
print("Cross Validation Model Comparison")

display(cv_table)
```

```python
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    brier_score_loss,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay,
    RocCurveDisplay,
    PrecisionRecallDisplay
)

from sklearn.tree import plot_tree
model = pruned_tree
y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)[:,1]
accuracy = accuracy_score(y_test,y_pred)

balanced_acc = balanced_accuracy_score(y_test,y_pred)
precision = precision_score(y_test,y_pred)
recall = recall_score(y_test,y_pred)
f1 = f1_score(y_test,y_pred)
roc = roc_auc_score(y_test,y_prob)
pr = average_precision_score(y_test,y_prob)
brier = brier_score_loss(y_test,y_prob)
tn,fp,fn,tp = confusion_matrix(y_test,y_pred).ravel()

specificity = tn/(tn+fp)

print("FINAL TEST PERFORMANCE")

print("Accuracy :",accuracy)

print("Balanced Accuracy :",balanced_acc)

print("Precision :",precision)

print("Sensitivity (Recall) :",recall)

print("Specificity :",specificity)

print("F1 Score :",f1)

print("ROC AUC :",roc)

print("PR AUC :",pr)

print("Brier Score :",brier)

print("\nClassification Report\n")

print(classification_report(y_test,y_pred))

plt.figure(figsize=(6,5))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    display_labels=["Benign","Malignant"],
    cmap="Blues"
)

plt.title("Confusion Matrix")

plt.show()
RocCurveDisplay.from_predictions(
    y_test,
    y_prob
)

plt.title("ROC Curve")

plt.show()

PrecisionRecallDisplay.from_predictions(
    y_test,
    y_prob
)

plt.title("Precision Recall Curve")

plt.show()
importance = pd.DataFrame({

    "Feature":X.columns,

    "Importance":model.feature_importances_

})

importance = importance.sort_values(
    by="Importance",
    ascending=False
)

print("\nTop Important Features\n")

display(importance.head(15))

plt.figure(figsize=(8,6))

sns.barplot(
    data=importance.head(10),
    x="Importance",
    y="Feature"
)

plt.title("Top 10 Feature Importance")

plt.show()
plt.figure(figsize=(18,10))

plot_tree(

    model,

    feature_names=X.columns,

    class_names=["Benign","Malignant"],

    filled=True,

    rounded=True,

    max_depth=3,

    fontsize=8

)

plt.title("Decision Tree")

plt.show()
locked_results = pd.DataFrame({

"Threshold":[0.50],
"TN":[tn],
"FP":[fp],
"FN":[fn],
"TP":[tp],
"Sensitivity":[recall],
"Specificity":[specificity],
"Precision":[precision],
"F1":[f1],
"ROC-AUC":[roc],
"PR-AUC":[pr],
"Brier":[brier]

})

print("SECTION 16.3")

display(locked_results)

tree_complexity = pd.DataFrame({

"Tree Version":[
"Original",
"Pruned"
],

"Criterion":[
"Gini",
"Gini"
],

"Depth":[
tree.get_depth(),
pruned_tree.get_depth()
],

"Leaves":[
tree.get_n_leaves(),
pruned_tree.get_n_leaves()
],

"Nodes":[
tree.tree_.node_count,
pruned_tree.tree_.node_count
],

"ccp_alpha":[
0,
0
],

"CV Score":[
tree_summary["test_accuracy"].mean(),
pruned_summary["test_accuracy"].mean()
],

"Interpretability":[
"Complex",
"Simpler"
]

})

print("SECTION 16.4")

display(tree_complexity)
robustness = pd.DataFrame({

"Group":["Overall Test Set"],

"n":[len(y_test)],

"Positive %":[round((y_test==1).mean()*100,2)],

"Sensitivity":[recall],

"Specificity":[specificity],

"Precision":[precision],

"Comment":["Overall evaluation only"]

})

print("SECTION 16.5")


display(robustness)

error_analysis = pd.DataFrame({

"Case Category":[

"False Negative",

"False Positive",

"Unstable Prediction",

"Out-of-Distribution"

],

"Observed Pattern":[

f"{fn} malignant cases missed",

f"{fp} benign cases predicted malignant",

"Tree complexity reduced after pruning",

"External datasets not evaluated"

],

"Possible Technical Explanation":[

"Borderline feature values",

"Overlap between classes",

"Original tree overfit",

"Distribution shift"

],

"Required Caution":[

"False negatives are clinically important",

"Extra diagnostic testing may occur",

"Prefer pruned tree",

"Do not generalize beyond dataset"

]

})

print("SECTION 16.6")

display(error_analysis)
```
